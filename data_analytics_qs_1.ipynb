```python
# Q2. Write a program to check whether a given year is a leap year.

year = 1700

if year % 400 == 0:
    print(year, "is a leap year")
elif year % 100 == 0:
    print(year, "is NOT a leap year")
elif year % 4 == 0:
    print(year, "is a leap year")
else:
    print(year, "is NOT a leap year")
```

```python
# Q5. Write a program to calculate the sum of first N natural numbers using a while loop.

n = 10
summ = 0
i = 1
while i <= n:
    summ += i
    i += 1
print("Sum of first", n, "natural numbers is", summ)
```

```python
# Q9. Write a function that takes the radius of a circle and returns its area.

def area_of_circle(radius):
    return 3.14 * radius ** 2

radius = 3
area = area_of_circle(radius)

print(f"Area of circle with radius {radius} units is {area} unit^2")
```

```python
# Q11. Write a function that takes a string and returns its length without using built-in length functions.

def str_length(s):
    ret = 0
    for c in s:
        ret += 1
    return ret

s = "helloworld"
print(f"Length of string \"{s}\" is {str_length(s)}")
```

```python
# Q12. Write a function that checks whether a number is prime.

def isprime(n):
    i = 2
    while i ** 2 <= n:
        if n % i == 0:
            return False
        i += 1
    return True

n = 13
if isprime(n):
    print(f"{n} is a prime number")
else:
    print(f"{n} is NOT a prime number")
```

```python
# Q13. Write a recursive function to find the nth Fibonacci number.

def fibo(n):
    if n == 1:
        return 0
    a = 0
    b = 1
    while n > 2:
        temp = b
        b = a + b
        a = temp
        n -= 1
    return b

n = 6
print(f"Fibonacci number at position {n} is {fibo(n)}")
```

```python
# Q14. Write a recursive function to reverse a string.

def recurse(original, result):
    if original == "":
        return result
    c = original[-1]
    result += c
    return recurse(original[:-1], result)

s = "Hello, World"
print(recurse(s, ""))
```

```python
# Q15. Write a program to find the sum of all numbers from 1 to n using a for loop.

def summ(n):
    ret = 0
    for i in range(1, n + 1):
        ret += i
    return ret

n = 4
print(f"Sum of all numbers from 1 to {n} is {summ(n)}")
```

```python
# Q16. Write a program to count how many even numbers are present in a list.

l = [324, 256, 346, 47, 68, 587, 567, 3542, 2654, 53, 84, 324]
count = 0
for n in l:
    count += (n + 1) % 2
print(f"There are {count} even numbers in the list")
```

```python
# Q17. Write a program to print all prime numbers between 1 and 100.

def isprime(n):
    i = 2
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

print("Prime Numbers From 1 to 100:")
for i in range(2, 101):
    if isprime(i):
        print(i, end = ' ')
```
