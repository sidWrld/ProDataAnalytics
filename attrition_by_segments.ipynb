```python
import pandas as pd
df = pd.read_csv("clean_data.csv")
```

```python
segments = ['Department', 'JobRole', 'Gender', 'MaritalStatus', 
            'OverTime', 'BusinessTravel', 'EducationField', 'JobLevel']

for seg in segments:
    print(f"\nAttrition % split by {seg}")
    
    # Percentage of Yes/No within each segment category (rows sum to 100%)
    pct = pd.crosstab(df[seg], df['Attrition'], normalize='index') * 100
    pct = pct.round(1)
    
    # Add count for context
    pct['Count'] = df[seg].value_counts()
    
    print(pct.sort_values('Yes', ascending=False))
```
