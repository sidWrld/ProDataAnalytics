<h2>Setting Up</h2>

```python
import pandas as pd
```

```python
#Reading data
df = pd.read_csv("data.csv")
```

```python
#Data Dictionary
print(df.info())
```

<h2>Data Cleaning</h2>

```python
#Checking for null values
df.isnull().sum()
```

```python
#Checking for duplicate rows
df.duplicated().sum()
```

```python
#Checking categorical values
print(df["Attrition"].unique())
print(df["Gender"].unique())
print(df["OverTime"].unique())
print(df["Over18"].unique())
```

```python
#Over18 only has single value, hence it is redundant and is to be removed
df = df.drop("Over18", axis = 1)
```

```python
#Removing other unneccesary columns
df = df.drop(["EmployeeCount", "StandardHours", "EmployeeNumber"], axis=1)
```

```python
df.info()
```

```python
#Analysing numeric values
df.describe()
```

```python
#Saving cleaned data
df.to_csv("clean_data.csv", index=False)
```

```python
# Numerical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
for i in numerical_columns:
    print(i, end=', ')
print('')
# Categorical columns
categorical_columns = df.select_dtypes(include=['object']).columns
for i in categorical_columns:
    print(i, end=', ')
```
